### OBJETOS INMUTABLES

![](/home/alumno/Documentos/codigo/python_fundamentals_notebook/imagenes/inmutabilidad.png)

Figura 6-3. Nombres y objetos después de ejecutar finalmente la asignación a = ‘spam’. Variable a hace referencia
el nuevo objeto (es decir, un trozo de memoria) creado al ejecutar la expresión literal ‘spam’, pero la variable b
todavía se refiere al objeto original 3. Debido a que esta asignación no es un cambio in situ en el objeto 3,
cambia sólo la variable a, no b.

```python
a = 3
b = a               # b es 3
print("b es ", b)

a = 3
b = a
a = 'spam'          # b es 3
print("b es ", b)
print("a es ", a)
```

### OBJETOS MUTABLES


---------------
#### Listas

```python
L = [2, 3, 4]	    # un objeto mutable
M = L			    # hace referencia al mismo objeto
L[0] = 24 		    # Un cambio in situ

print("L = ", L)    # [24, 3, 4]
print("M = ", M)    # [24, 3, 4]
```

-------------------------------------
Copia de objetos en vez de referencia:

```python
L = [2, 3, 4]
M = L[:]		    # Hces la copia de L (or list(L), copy.copy(L), etc.)
L[0] = 24

print("L = ", L)    # [24, 3, 4]

print("M = ", M)    # [2, 3, 4]  M no a cambiado
```

### Compartir referncias y igualdad

- Operador **==** igualdad de valores.
- Operador **is** identidad de objetos.

```python
L = [1, 2, 3]
M = L 			    # M and L hacen referencia al mimo objeto

L == M 			    # True: mismos valores  => operador igualdad de valores
L is M 			    # True: operador identidad de objetos: compara las referencias (los punteros)
```

```python
L = [1, 2, 3]
M = [1, 2, 3]	    # M and L referencian a diferentes objetos con los mismos valores

print(
    "L == M",
    L == M 			# True: los mismos valores 
)

print(
    "L is M", 	    # False: operador identidad de objetos: compara las referencias (los punteros)
    L is M
)
```

### CACHE
Sin embargo, debido a que los números enteros y cadenas pequeños se almacenan en caché y se reutilizan, nos dice que hacen referencia al mismo objeto único.

```python
X = 42
Y = 42		    # Deberian ser dos objetos diferentes 

X  == Y         # True
X is Y 		    # True: De todos modos, el mismo objeto: ¡¡almacenamiento en caché en el trabajo!!
```

Averiguar el numero de refencias a un objeto:

```python
import sys
sys.getrefcount(1)		# Número de punteros a esta pieza de memoria compartida
```
