## GLOBAL SCOPE
- Los nombres globales son variables asignadas en el nivel superior del archivo del módulo adjunto.
- Los nombres globales deben declararse sólo si están asignados dentro de una función.
- Global names may be referenced within a function without being declared.

```python
x = 88 		# Global X

def func():
	x = 99 	# Local X: se esconde global, pero queremos esto aquí

func()
print(x) #	 Prints 88: unchanged
```

```python
x = 88 			# Global X

def func():
	global x
	x = 99 		# Global X: fuera de def

func()
print("x =", x) # Prints 99
```

-------------
## Regla LEGB

```python
# Se puede hacer referencia a nombres globales dentro de una función sin declararlos.

y, z = 1, 2         # Variables globales en el módulo

def all_global():
	global x        # Declarar globales asignados
	x = y + z       # No es necesario declarar y, z: regla LEGB 
                    # x existe ahora en el ambito global con valor 3

all_global()
print('x = %d, y = %d, z = %d' % (x, y, z))
```

## NESTED SCOPES

```python
x = 99                      # Global scope name: no usado

def f1():
    x = 88
    def f2():               # encerrando def local
        print("x local = ", x)    # Referencia realizada en def anidada
    f2()                    # f2 es una función temporal que vive sólo durante la ejecución        de (y es visible solo para codificar en) el f1 que lo encierra

f1()                        # Prints 88: encerrando def local
print("x global = ", x)     # salida: X = 88; X sigue valiendo 99

# f2()  NameError: name 'f2' is not defined  No se puede invocar a f2() desde el modulo principal
```

## Funciones de fábrica:Closures(cierres)

Un cierre o una función de fábrica, el primero describe una técnica de programación funcional y el segundo denota un patrón de diseño. El objeto de función en cuestión recuerda valores en ámbitos adjuntos independientemente de si esos ámbitos todavía están presentes en la memoria.

```python
def f1():
	x = 88                 # enclosing scope
	def f2():
		print("x =", x)    # Recuerda x al encerrar el ámbito def
	return f2 	           # Devuelve f2 pero no lo llames => f1() devuelve el objeto funcion con                              nombre (referencia) f2

action = f1()       # Make, return function => action es ahora una función: action = f2
action() 			# Call it now: prints 88  == f2()

# Las funciones son objetos en Python como todo lo demás y pueden devolverse como valores de retorno desde otras funciones. 
# Lo más importante es que f2 recuerda la x del ámbito que lo encierra en f1, aunque f1 ya no esté activo.
```

```python
def maker(n):
	def action(x):		# Make and return action
		return x ** n 	# action retains n from enclosing scope
	return action

f = maker(2)            # Pass 2 to argument n -> return x ** 2
f(3)                    # Pass 3 to x, n remembers "2" -> return 3 ** 2

f(4)		            # return 4 ** 2
```
