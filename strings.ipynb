*Secuencia de Operaciones*
---------------------
***Como secuencias, las cadenas admiten operaciones que suponen un orden posicional entre ellas
artículos.***

```python
S = 'Spam'
len(S)  
```

***En Python, también podemos indexar hacia atrás, desde el final—recuento de índices positivos desde
a la izquierda, y los índices negativos cuentan hacia atrás desde la derecha:***

```python
S[-1]
```

*Inmutabilidad*
---------------
***Cada operación de cadena está definida para producir una nueva
cadena como resultado, debido a que las cadenas son inmutables en Python—no se pueden cambiar
en su lugar después de su creación.***

```python
S
```

```python
S[0] = 'z' #Los objetos nopueden ser modificados, por lo que no se puede cambiar un caracter de un string.
```

```python
S = 'z' + S[1:]
S
```

***Estrictamente hablando, puedes cambiar los datos basados en texto si los expandes a un
lista de personajes individuales y volver a unirlos sin nada intermedio, o usar el
tipo bytearray más nuevo***

```python
S = 'shrubbery'
L = list(S)
L 
```

```python
L[1] = 'c'
''.join(L)
```

*Métodos específicos de tipo*
-----------------------------
***Las cadenas también tienen operaciones todas
propios, disponibles como métodos—funciones que están asociadas a un específico y actúan sobre él
objeto, que se activan con una expresión de llamada.***

***Por ejemplo, el método de búsqueda (find) de cadenas es la operación básica de búsqueda de subcadenas (devuelve
el desplazamiento de la subcadena pasada, o −1 si no está presente)***

```python
S = 'Spam'
S.find('pa')
```

```python
S
```

```python
S.replace('pa', 'XYZ')
```

```python
S
```

***Nuevamente, a pesar de los nombres de estos métodos de cadena, no estamos cambiando el original
cadenas aquí, pero creando nuevas cadenas como resultado —porque las cadenas son inmutables,
esta es la única manera en que esto puede funcionar***

Otros métodos dividen una cadena en subcadenas en un delimitador (útil como
una forma sencilla de análisis)

```python
line = 'aaa,bbb,ccccc,dd'
line.split(',')
```

```python
S = 'spam'
S.upper() # Conversiones de mayúsculas y minúsculas
```

```python
line = 'aaa,bbb,ccccc,dd\n'
line.rstrip() # Elimine los caracteres de espacio en blanco en el lado derecho
```

```python
line.rstrip().split(',') # combina dos operaciones
```

***Los strings también admiten un modo avanzado
operación de sustitución conocida como formato, disponible como expresión (el original-
inal) y una llamada a un método de cadena (nuevo a partir de 2.6 y 3.0); el segundo de estos le permite
para omitir los números de valores de argumentos relativos***

```python
'%s, eggs, and %s' % ('spam', 'SPAM!') # Formateo de expresión 
```

```python
'{0}, eggs, and {1}'.format('spam', 'SPAM!') # Formateo de método 
```

```python
'{}, eggs, and {}'.format('spam', 'SPAM!') # números opcionales 
```

*Listas*
--------
***Las listas son colecciones ordenadas posicionalmente de objetos tipificados arbitrariamente y no tienen un tamaño fijo***

*Secuencia de operaciones*

```python
L = [123, 'spam', 1.23] # Una lista de tres objetos de diferentes tipos
len(L)  # Número de elementos en la lista
```

```python
L[0] #Indexadas por posicion
```

```python
L[:-1] # Al cortar una lista se devuelve una nueva lista
```

```python
L + [4, 5, 6] #Concatenar/repetir para hacer nuevas lisats tambien
```

```python
L * 2
[123, 'spam', 1.23, 123, 'spam', 1.23]
```

```python
L # No cambaimos la lista original
```

```python
*Tipos especificos de operaciones*
```

```python
L.append('NI') # Creciendo: agregar objeto al final de la lista
L
```

```python
L.pop(2) # Reducción: elimina un elemento en el medio
```

```python
L # "del L[2]" también se elimina de una lista
```

*Nido*
***Una característica interesante de los tipos de datos principales de Python es que admiten anidamiento arbitrario—nosotros
Podemos anidarlos en cualquier combinación y tan profundamente como queramos.***

```python
M = [[1, 2, 3], # Una matriz 3 × 3, como listas anidadas
     [4, 5, 6],
     [7, 8, 9]] # El código puede abarcar líneas si está entre corchetes
M
```

```python
M[1] #Obtener la fila 2
```

```python
M[1][2] # Obtenga la fila 2, luego obtenga el elemento 3 dentro de la fila
```

*Comprensiones*
***Además de las operaciones de secuencia y los métodos de lista, Python incluye un método más avanzado
operación conocida como expresión de comprensión de listas, que resulta ser poderosa
forma de procesar estructuras como nuestra matriz.***

```python
col2 = [row[1] for row in M] #Colecciona los elementos de la columna 2 
col2
```

```python
M #La matriz no cambia
```
