## 1. Argumentos y referencias compartidas

### Paso de objetos inmutables:

```python
def f(a):	    	# a is assigned to (references) the passed object
	a = 99 			# Changes local variable a only

b = 88		
f(b)				# a and b both reference same 88 initially
print("b = ", b)	# b is not changed
```

----------------
### Paso de objetos mutables:

```python
def changer(a, b):	    # Arguments assigned references to objects
		a = 2			# Changes local name's value only
		b[0] = 'spam' 	# Changes shared object in place
	
					
x = 1
l = [1, 2]
changer(x, l)           # Caller: Pass immutable and mutable objects
print("l = ", l)        # ['spam', 2]  # x is unchanged, l is different!
assert x == 1, "x es 1"
```

---------------------
### Cómo evitar que una rutina modifique los elementos mutables:
- Pasar una copia

```python
l = [1, 2]
changer(x, l[:]) 	# Pass a copy, so our 'l' does not change 

print("l = ", l) 
```

- Utilizar invariantes (en las estructuras de datos de entrada)

```python
l = [1, 2]
def changer(a, b):
    b = b[:]		# Copy input list so we don't impact caller
    a = 2
    b[0] = 'spam'	# Changes our list copy only

changer(x, l)
assert l == [1, 2], "l es una invariante"
print("l = ", l) 
```

- Convertir el objeto mutable a un objeto inmutable:

```python
l = [1, 2]
def changer(a, b):	   
		a = 2			
		b[0] = 'spam'   # TypeError: 'tuple' object does not support item assignment

changer(x, tuple(l))	# Pass a tuple, so changes are errors

print("l = ", l)
assert l == [1, 2], "l es una invariante"
```

### 2 Modos especiales de coincidencia de argumentos
#### ver figura 06 en Drive
Tenga cuidado de no confundir la sintaxis especial name=value en un encabezado de función y una llamada de función; en la llamada significa un argumento de palabra clave de coincidencia por nombre, mientras que en el encabezado especifica un valor predeterminado para un argumento opcional.

### Palabras clave y ejemplos predeterminados
#### 1. Comportamiento por defecto:

```python
def f(a, b, c): 
    print(a, b, c)

f(1, 2, 3)          # 1 2 3
```

---------------------------
#### 2.Palabras clave
El orden de izquierda a derecha de los argumentos ya no importa cuando se utilizan palabras clave porque los argumentos coinciden por nombre, no por posición.

```python
def f(a, b, c): 
    print(a, b, c)

f(c=3, b=2, a=1)    # 1 2 3
```

```python
f(1, c=3, b=2)		# a gets 1 by position, b and c passed by name
                    # 1 2 3

# they make your calls a bit more self documenting
# f(name='Bob', age=40, job='dev')
```

----------------------------
#### 3.Predeterminados
Los valores predeterminados nos permiten hacer que los argumentos de la función seleccionada sean opcionales; si no se pasa un valor, al argumento se le asigna su valor predeterminado antes de que se ejecute la función.

```python
def f(a, b=2, c=3): 	# a required, b and c optional
	print(a, b, c)


f(1)        # 1 2 3
f(a=1)      # 1 2 3
```
